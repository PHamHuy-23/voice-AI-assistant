# REPORT 2.4 — FEATURE EXTRACTION

Trích xuất và minh họa các đặc trưng tiếng nói: Waveform, STFT, Log Mel-Spectrogram, MFCC, RMS Energy, Zero Crossing Rate.


```python
from src.features.audio_features import FeatureExtractor
from src.features.visualization import plot_feature_gallery
from src.utils.config import load_yaml
import numpy as np

model_cfg = load_yaml('configs/model/td_resnet.yaml')
extractor = FeatureExtractor(model_cfg)

# Tao tin hieu audio mau (1s sinewave 16kHz)
sr = 16000
t = np.linspace(0, 1.0, sr, endpoint=False)
audio = 0.5 * np.sin(2 * np.pi * 440 * t) + 0.1 * np.random.randn(sr)
audio = audio.astype(np.float32)

spec = extractor.compute_linear_spectrogram(audio)
mel = extractor.compute_mel_spectrogram(audio)
mfcc = extractor.compute_mfcc(audio)
energy = extractor.compute_rms_energy(audio)
zcr = extractor.compute_zero_crossing_rate(audio)

plot_feature_gallery(audio, spec, mel, mfcc, energy, zcr, sr, title='Report 2.4: Feature Gallery', save_path='reports/figures/feature_gallery.png')
print('Saved figure to: reports/figures/feature_gallery.png')
```
